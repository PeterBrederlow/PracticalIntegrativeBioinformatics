# DataFrame aggregation

## Count the number of characters from each country

## Calculate the average weight and age of characters by species
Store the result in `df_characters_avg`. The new dataframe should only contain the columns `species`, `avg_weight` and `avg_age`.

```python
# YOUR CODE HERE
df_characters_avg = df_characters.groupby('species', as_index=False).agg(
    avg_weight=('weight', 'mean'),
    avg_age=('age', 'mean'))
df_characters_avg
```

```python
assert 'df_characters_avg' in locals()
assert 'avg_weight' in df_characters_avg.columns
assert 'avg_age' in df_characters_avg.columns
```

## List the heaviest characters per country
Store the result in `df_chars_heaviest`. The new dataframe should only contain the columns `country`, `name`, `weight`.

```python
# YOUR CODE HERE
heaviest_indices = df_characters.groupby('country')['weight'].idxmax()
df_chars_heaviest = df_characters.loc[heaviest_indices, ['country', 'name', 'weight']]
df_chars_heaviest
```

# Joining (or merging) DataFrames
We now have three dataframes: `df_species`, `df_characters` and `df_countries`. Take another look at all of them and try to come up with a way to combine their information.
Create another dataframe called `df_merge` including all three dataframes. There are some things you should consider, when you create your new dataframe:
1. We are going to use our characters from `df_characters` as main target, so leave out all countries or species without one.
2. Character and species each have weights, use `weight_indiv` for a characters weight and `weight_species` for the species weight.
3. Countries have a column `country` and `code`, think about which you should use as a key to join your dataframes.
4. Try to solve this task using (mostly) the `merge` function in pandas.

<div class="alert alert-block alert-info"><b>Hint: Should you struggle or just want to compare your result, you can take a look at the merged dataframe <a href=https://github.com/Practical-Integrative-Bioinformatics/Introduction/blob/main/example_data/day1.tsv>here</a>.</</a></div> Watch out, as this is not going to look exactly like your dataframe, as this contains additional columns and different ages.

```python
# YOUR CODE HERE
#inner merge, since only countries and species that are in df_characters are kept
df_merge = pd.merge(df_characters, df_countries, left_on='country', right_on='code', how = 'left')
df_merge
df_merge = pd.merge(df_merge, df_species, left_on='species', right_on='species', how = 'left')
df_merge
```

```python
assert 'df_merge' in locals()
```
